# Chapter 6: k-means on outlets and recipes

The software finds the groups. You name them. Synthetic data only; no real brewery's numbers.

What to look at: the group sizes and the centres, then change k and see what moves.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

o = pd.read_csv('../data/outlets.csv')
F = ['monthly_volume_hl', 'draught_share_pct', 'sku_count', 'weekend_share_pct', 'na_share_pct']
x = StandardScaler().fit_transform(o[F])  # scale first, or volume swamps the shares
o['group'] = KMeans(3, n_init=10, random_state=0).fit_predict(x)
print(o.group.value_counts())
print(o.groupby('group')[F].mean().round(1))
```

```python
plt.scatter(o.monthly_volume_hl, o.weekend_share_pct, c=o.group, s=8)
plt.xlabel('Monthly volume (hl)'); plt.ylabel('Weekend share of sales (%)')
plt.show()
```

## Try other values of k

k is a choice, not a fact.

```python
for k in (2, 3, 4, 5):
    sizes = KMeans(k, n_init=10, random_state=0).fit_predict(x)
    print(k, sorted(pd.Series(sizes).value_counts(), reverse=True))
```

## 150 recipes, five groups

Name each group from its centre: pale and light, hoppy and gold, amber and malty, dark and strong, odd experiments.

```python
r = pd.read_csv('../data/recipes.csv')
R = ['og_plato', 'ibu', 'colour_ebc', 'abv_pct']
r['group'] = KMeans(5, n_init=10, random_state=0).fit_predict(StandardScaler().fit_transform(r[R]))
print(r.groupby('group')[R].mean().round(1).assign(recipes=r.group.value_counts()))
```
