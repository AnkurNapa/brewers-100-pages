# Chapter 6: the odd fermentation curve

200 healthy curves of one lager, a reading every 15 minutes for six days. Learn the normal band, then check 40 new curves. Synthetic data only; no real brewery's numbers.

What to look at: which curves leave the band and what inspection found.

```python
import pandas as pd
import matplotlib.pyplot as plt

h = pd.read_csv('../data/ferm_healthy.csv').pivot(index='curve_id', columns='reading', values='plato')
n = pd.read_csv('../data/ferm_new.csv').pivot(index='curve_id', columns='reading', values='plato')
print('Healthy curves:', h.shape[0], 'readings each:', h.shape[1], 'in all:', h.size)
```

Normal is the mean at each reading plus or minus four standard deviations. A curve is flagged when more than 5 percent of its readings sit outside.

```python
mu, sd = h.mean(), h.std()
outside = ((n - mu).abs() > 4 * sd).mean(axis=1)
flags = outside[outside > 0.05]
print(flags.round(2))
causes = pd.read_csv('../data/ferm_new_inspection.csv').set_index('curve_id')
print(causes.loc[flags.index])
```

```python
days = h.columns / 96
plt.fill_between(days, mu - 4 * sd, mu + 4 * sd, alpha=0.3)
for cid, row in n.iterrows():
    plt.plot(days, row, lw=0.6, color='red' if cid in flags.index else 'grey')
plt.xlabel('Days'); plt.ylabel('Gravity (P)')
plt.show()
```
