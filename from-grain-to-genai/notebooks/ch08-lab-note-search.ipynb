# Chapter 8: search lab notes by meaning

Keyword search misses notes that use other words. Here each note becomes a short vector of numbers (latent semantic analysis: TF-IDF then SVD), a small stand-in for the embeddings in the chapter. Swap in an embedding model if you have one. Synthetic data only; no real brewery's numbers.

What to look at: which notes each search finds.

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import TruncatedSVD
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import normalize

notes = pd.read_csv('../data/lab_notes.csv')
vec = TfidfVectorizer(stop_words='english')
svd = TruncatedSVD(4, random_state=0)
Z = normalize(svd.fit_transform(vec.fit_transform(notes.note)))

def search(q, top=5):
    score = Z @ normalize(svd.transform(vec.transform([q]))).ravel()
    return notes.iloc[np.argsort(-score)[:top]][['note_date', 'vessel', 'note']]
```

```python
print(notes[notes.note.str.contains('diacetyl', case=False)][['note_date', 'note']])  # keyword
print(search('diacetyl'))  # meaning
```

```python
print(search('fermentation stalled at 1.020 after a cold crash'))
```

Every result carries its date and vessel, so a brewer reads the note and judges it.
