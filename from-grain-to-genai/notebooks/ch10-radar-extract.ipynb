# Chapter 10: count the radar extract

An anonymised extract of the author's Beverage-AI Radar as at 15 September 2026: one row per entry with its theme, grade and whether it mentions generative AI. No company names. The grade and the shipping and no claim counts by theme match the published analysis; the pilot, research and ungraded split inside each theme is allocated for this extract.

What to look at: a theme where your plant has data and few vendors ship.

```python
import pandas as pd

r = pd.read_csv('../data/radar_extract.csv')
print(len(r), 'entries')
print(r.grade.value_counts())
```

```python
print(pd.crosstab(r.theme, r.grade, margins=True))
```

```python
share = r.assign(pilot_or_research=r.grade.isin(['pilot', 'research'])).groupby('theme').pilot_or_research.mean()
print((100 * share).round(0).sort_values(ascending=False))
print('Mention generative AI:', r.mentions_genai.sum())
```
