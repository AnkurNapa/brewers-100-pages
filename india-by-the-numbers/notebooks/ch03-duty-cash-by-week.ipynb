# Chapter 3.4: cash tied up in duty by week

Kestrel pays INR 1,096.68 of excise and pass fee on each case at dispatch (State S, synthetic).
A wholesaler repays 30 days later; the corporation variant pays 60 days after receipt (transit days are ignored here).
Look at the peak, not the average: the peak is what your bank line must cover.

```python
import sys
sys.path.append('..')
import pandas as pd
import matplotlib.pyplot as plt
from calculator import inr, r2
```

```python
DUTY_AND_FEE = 1081.08 + 15.60   # a 12 x 650 ml case, chapter 5
plan = pd.read_csv('../data/dispatch_plan_weekly.csv', parse_dates=['week_start'])
plan['duty_paid'] = plan.cases * DUTY_AND_FEE
days = pd.date_range(plan.week_start.min(), plan.week_start.max() + pd.Timedelta(days=70))
def tied_up(lag_days):
    paid = plan.set_index('week_start').duty_paid
    return pd.Series([paid[(paid.index <= d) & (paid.index > d - pd.Timedelta(days=lag_days))].sum() for d in days], index=days)
tied = pd.DataFrame({'30 days': tied_up(30), '60 days': tied_up(60)})
weekly = tied.resample('W-MON').max()
print('peak cash tied up in duty, INR:', inr(weekly['30 days'].max(), 0), 'at 30 days;', inr(weekly['60 days'].max(), 0), 'at 60 days')
```

```python
ax = (weekly / 1e5).plot(figsize=(8, 3), title='Duty and fee paid but not yet repaid (INR lakh)')
ax.set_ylabel('INR lakh'); plt.show()
```

## The chapter's per case figures

```python
carry_30 = r2(DUTY_AND_FEE * 0.12 * 30 / 365)
carry_60 = r2(DUTY_AND_FEE * 0.12 * 60 / 365)
front_1000 = r2(1000 * DUTY_AND_FEE)
print('carrying cost a case:', carry_30, 'at 30 days;', carry_60, 'at 60 days')
print('duty and fee fronted on 1,000 cases: INR', inr(front_1000))
```
