# Chapter 6.5: a duty and MRP calculator in 20 lines

State S rules (synthetic): duty is the larger of 110 percent of DMC and INR 50 a bulk litre; pass fee INR 2 a bulk litre;
wholesaler 8 percent; retailer 15 percent; round up to the next INR 5 a bottle. Change the pack, strength or DMC range.
Find the DMC at which your MRP drops one rounding step: that is your cliff.

```python
import math

def mrp(dmc, bottles=12, litres=0.65, share=1.10, floor=50, fee=2, buyer=0.08, retail=0.15, step=5):
    bl = bottles * litres
    duty = round(max(share * dmc, floor * bl), 2)
    landed = round(dmc + duty + fee * bl, 2)
    wholesale = round(landed * (1 + buyer), 2)
    shelf = round(wholesale * (1 + retail), 2)
    bottle = shelf / bottles
    return round(bottle, 2), step * math.ceil(round(bottle, 6) / step)

for dmc in range(940, 1061, 10):
    before, price = mrp(dmc)
    print(f'DMC {dmc:5d}  before rounding {before:7.2f}  MRP {price}')

lager = 12600 * 0.078
low, high = mrp(lager * 0.95), mrp(lager * 1.05)
print('Kestrel Lager:', mrp(lager), '; 5 percent cut:', low, '; 5 percent rise:', high)
```
