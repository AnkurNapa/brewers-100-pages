Counts from the author's review of 116 beer AI vendors, public websites, researched 24 September 2026. Counts only: no vendor is named and no row identifies one. They will move.

Add your own shortlist below and place each offer in one of the five ways.

```python
import pandas as pd
import matplotlib.pyplot as plt

counts = pd.read_csv("../data/vendor_counts.csv")
counts
```

## Figure 3.2: from claim to working to proof

```python
c = counts.set_index("measure")["count"]
steps = [("Claim a number", c["Make a quantified claim"]),
         ("Also explain any part", c["Of those 23 also make a quantified claim"]),
         ("Explain it properly", c["Of those 3 also make a quantified claim"])]
fig, ax = plt.subplots(figsize=(6, 2.4))
ax.barh([s for s, _ in steps][::-1], [n for _, n in steps][::-1], color=["#0b4f45", "#2f7f72", "#cfe6e1"][::-1])
for i, (_, n) in enumerate(steps[::-1]):
    ax.text(n + 0.5, i, str(n), va="center")
ax.set_xlabel("Vendors, of 116")
plt.tight_layout()
plt.show()
```

## Your shortlist

The five ways, weakest first: headline percentage, named case, calculator, independent study, shared risk.

```python
WAYS = ["headline percentage", "named case", "calculator", "independent study", "shared risk"]
shortlist = pd.DataFrame({"offer": ["Shortlist 1", "Shortlist 2", "Shortlist 3"],
                          "way": ["headline percentage", "calculator", "named case"]})
shortlist["rank"] = shortlist["way"].map({w: i + 1 for i, w in enumerate(WAYS)})
shortlist.sort_values("rank", ascending=False)
```
