# Claim parts (chapters 1 and 2)

Chapter 1 asks you to turn a claim into a table of its missing parts. Chapter 2.5 asks which of five words never appear: baseline, period, measured, net, deducted.

Paste a claim into the cell below. The example is the chapter 1 email, written for the book. All data is synthetic.

```python
import re
import pandas as pd

claim = """Our platform cuts cellar beer loss by 20 percent. Customers see payback in months.
Book a demo and we will send pricing."""
```

## The five words

A word that never appears is a question to send the vendor.

```python
WORDS = ["baseline", "period", "measured", "net", "deducted"]
words = pd.DataFrame({"word": WORDS,
                      "appears": [bool(re.search(rf"\b{w}\b", claim, re.I)) for w in WORDS]})
words
```

## The five parts

Each part gets a rough text test. Read the claim yourself too: the test only finds words, not meaning.

```python
PARTS = {
    "Number and unit": r"\d+(\.\d+)?\s*(percent|%|points?|hL|USD)",
    "Baseline": r"baseline|against|versus|compared with",
    "Period": r"\b20\d\d\b|twelve months|over the last|a year|per year",
    "Who measured": r"measured|audited|verified|counted by",
    "Deducted": r"\bnet\b|deducted|after the (fee|licence)",
}
table = pd.DataFrame([(p, "found" if re.search(rx, claim, re.I) else "missing") for p, rx in PARTS.items()],
                     columns=["Part", "What the claim says"])
missing = (table["What the claim says"] == "missing").sum()
print(table.to_string(index=False))
print(f"\n{missing} of 5 parts missing.")
print("Three or more missing: send a written request for working, not a demo." if missing >= 3 else "Score it next (chapter 4).")
```
