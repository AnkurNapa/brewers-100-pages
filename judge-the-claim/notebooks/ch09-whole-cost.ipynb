# whole-cost (chapter 9)

Licence, cloud bill, install, data preparation and people time, added over the years, then the exit. All figures synthetic, USD. Change the inputs to your own quote.

```python
import pandas as pd
inp = pd.read_csv("../data/kestrel_inputs.csv").set_index("item")["value"]
inp
```

```python
years = 3
licence = 12 * inp["licence_usd_a_month"]
cloud = 12 * inp["cloud_bill_usd_a_month"]
install = inp["install_usd"]
prep = inp["data_prep_hours"] * inp["contractor_rate_usd_an_hour"]
rate = inp["loaded_rate_usd_an_hour"]
people = [(13 * inp["people_hours_a_week_first_13_weeks"] + 39 * inp["people_hours_a_week_after"]) * rate] + \
         [52 * inp["people_hours_a_week_after"] * rate] * (years - 1)

table = pd.DataFrame({
    "Licence": [licence] * years,
    "Cloud bill": [cloud] * years,
    "Install, one off": [install] + [0] * (years - 1),
    "Data preparation, one off": [prep] + [0] * (years - 1),
    "Brewery people time": people,
}, index=[f"Year {y + 1}" for y in range(years)]).T
table.loc["Whole cost"] = table.sum()
table
```

## Add the years and the exit (9.4)

The exit lands in the year the contract ends.

```python
total = table.loc["Whole cost"].sum()
exit_cost = inp["exit_hours"] * inp["contractor_rate_usd_an_hour"]
print(f"Whole cost over {years} years: {total:,.0f}")
print(f"Exit: {exit_cost:,.0f}; with the exit: {total + exit_cost:,.0f}")
claim_at_30 = inp["volume_hL_a_year"] * inp["cellar_loss_rate"] * inp["share_recovered_claim"] * inp["value_lost_hL_replacement_usd"]
print(f"Full claim at replacement cost: {claim_at_30:,.0f} a year against {table.loc['Whole cost', 'Year 2']:,.0f} a year after year one")
```
