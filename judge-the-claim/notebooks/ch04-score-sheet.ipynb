# The Working Score sheet (chapter 4)

Two people score each claim on their own, one point per question. This notebook reads their scores from a CSV and flags any claim where the two totals differ by more than one point. It lists the questions they split on, so you know which page to read together.

The example is claims A, B and C of chapter 4.4, constructed for the book. The owner's points match the book; the head brewer's are a synthetic second opinion.

```python
import pandas as pd

s = pd.read_csv("../data/ch04_scores.csv")
totals = s.pivot_table(index="claim", columns="scorer", values="point", aggfunc="sum")
totals
```

## Read the score (chapter 4.3)

```python
def band(score):
    return "Discount to zero" if score <= 2 else ("Ask for working" if score <= 5 else "Worth a pilot")

totals["owner band"] = totals["owner"].map(band)
totals["gap"] = (totals["owner"] - totals["head brewer"]).abs()
totals["read together"] = totals["gap"] > 1
totals
```

## Where the scorers split

```python
wide = s.pivot_table(index=["claim", "question", "text"], columns="scorer", values="point").reset_index()
split = wide[wide["owner"] != wide["head brewer"]]
split[split["claim"].isin(totals.index[totals["read together"]])]
```
