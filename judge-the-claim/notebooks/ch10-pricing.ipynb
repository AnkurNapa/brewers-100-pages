# Fixed fee or outcome fee (chapter 10)

Price the same quote two ways on your own verified results. The outcome fee is a share of the verified saving, valued at a rate you agree (here the USD 110 upper bound, chapter 7.1). Synthetic figures, USD.

## The same results, priced two ways (10.2)

```python
import pandas as pd
import matplotlib.pyplot as plt

fixed_fee, share, value = 16_800, 0.33, 110
results_hL = [100, 300, 480]   # 0.1, 0.3 and 0.48 points of 100,000 hL
t = pd.DataFrame({"hL": results_hL})
t["saving"] = t["hL"] * value
t["fixed: net"] = t["saving"] - fixed_fee
t["outcome fee"] = (t["saving"] * share).round(0)
t["outcome: net"] = t["saving"] - t["outcome fee"]
t["saving at USD 30"] = t["hL"] * 30
t
```

## Figure 10.1

```python
fig, ax = plt.subplots(figsize=(6, 3.2))
x = range(len(t))
ax.bar([i - 0.2 for i in x], t["fixed: net"], 0.4, label="Fixed fee")
ax.bar([i + 0.2 for i in x], t["outcome: net"], 0.4, label="Outcome fee")
ax.set_xticks(list(x), [f"{h} hL" for h in results_hL])
ax.axhline(0, color="grey", lw=0.8)
ax.set_ylabel("Net to brewery, USD a year")
ax.legend()
plt.tight_layout()
plt.show()
```
