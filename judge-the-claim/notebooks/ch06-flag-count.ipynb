# Flag count from the first call (chapter 6)

Write each vendor's answers as flags present (1) or absent (0). The notebook counts red and green flags for each vendor. Two constructed vendors are filled in as an example.

```python
import pandas as pd

answers = pd.read_csv("../data/ch06_first_call.csv")
count = answers.pivot_table(index="vendor", columns="flag_type", values="present", aggfunc="sum")
count["reading"] = count["red"].map(lambda r: "has not tried to be checked" if r >= 7 else ("normal" if r <= 3 else "push hard"))
count
```
