Ten synthetic questions, each with the controlled document and revision that answers it, plus what an assistant cited. The notebook counts answers with no source and answers from a superseded revision. No language model is called: paste your own assistant's citations into the CSV.

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
DATA = "../data/"
```

```python
q = pd.read_csv(DATA + "ch12_sop_testset.csv", keep_default_na=False)
q["no source"] = (q.answered == "yes") & (q.cited_doc == "")
q["right document"] = q.cited_doc == q.answer_doc
q["superseded revision"] = q["right document"] & (q.cited_rev != q.answer_rev)
q["said it cannot find it"] = q.answered == "no"
q[["question", "cited_doc", "cited_rev", "no source", "superseded revision", "said it cannot find it"]]
```

```python
print("answers with no source:", q["no source"].sum())
print("answers from a superseded revision:", q["superseded revision"].sum())
print("right document and revision:", (q["right document"] & ~q["superseded revision"]).sum(), "of", len(q))
print("honest 'I cannot find this':", q["said it cannot find it"].sum())
```

**What to look at.** An answer with no source is a failure even when the words are right. A superseded revision reads just as calmly as the current one. Rerun the set whenever a document changes.
