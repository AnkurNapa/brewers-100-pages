# Chapter 1: the synthetic week, counted by stream
The chapter says a lab of 60 samples a day makes 420 samples a week and well over a thousand results. This notebook loads the week and counts results per stream.

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
DATA = "../data/"
```

```python
samples = pd.read_csv(DATA + "week_sample.csv")
results = pd.read_csv(DATA + "week_result.csv")
tests = pd.read_csv(DATA + "week_test.csv")
print(len(samples), "samples;", len(results), "results")
print(samples.sample_id.str[1:7].value_counts().sort_index())
```

```python
by_stream = results.merge(tests[["test_code", "stream"]], on="test_code", how="left")
by_stream["stream"] = by_stream["stream"].fillna("test code not in the test table")
by_stream.stream.value_counts()
```

```python
per_day = samples.groupby("stage").size() / 7
print(per_day.sort_values(ascending=False))
print("micro and hygiene samples a day:", per_day["process micro"] + per_day["hygiene"])
```

```python
limit = {"MIC_MEM": 5, "MIC_PLATE": 10, "MIC_RINSE": 10, "ATP": 301}
micro = results[results.test_code.isin(limit)].copy()
micro["at_action"] = [float(v) >= limit[c] for v, c in zip(micro.value, micro.test_code)]
print(micro.at_action.sum(), "of", len(micro), "micro and hygiene results reach the action limit:", round(100 * micro.at_action.mean(), 1), "percent")
```

**What to look at.** Which streams carry most rows. How few micro rows reach the action limit. The one row whose test code is not in the test table is a planted fault; chapter 3 finds it again.
