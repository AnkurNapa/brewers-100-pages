# Chapter 5: individuals and moving range chart on bright beer haze
120 synthetic haze results at 90 degrees on Kestrel lager, seed 5, with a planted shift of 0.06 EBC from result 81. Limits come from the first 60.

**No code tier.** Put the 120 results in column A. In B2 type =ABS(A2-A1) and fill down. Average the first 60 of each column. Draw lines at the average plus and minus 2.66 x the average moving range. Mark points beyond a line and runs of eight on one side.

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
DATA = "../data/"
```

```python
x = pd.read_csv(DATA + "ch05_haze.csv").haze90_ebc.to_numpy()
base = x[:60]
cl = base.mean(); amr = np.abs(np.diff(base)).mean()
ucl, lcl, mr_ucl = cl + 2.66 * amr, cl - 2.66 * amr, 3.267 * amr
print(f"centre {cl:.3f}  average moving range {amr:.4f}  limits {lcl:.3f} to {ucl:.3f}  MR limit {mr_ucl:.3f}  sigma {amr / 1.128:.3f}")
```

```python
beyond = [i + 1 for i, v in enumerate(x) if v > ucl or v < lcl]
run, side, eighth = 0, 0, None
for i, v in enumerate(x, 1):
    s = 1 if v > cl else -1
    run = run + 1 if s == side else 1; side = s
    if run == 8 and eighth is None: eighth = i
print("points beyond a limit:", beyond, "| eighth in a row at", eighth)
print(f"mean of 81 to 120: {x[80:].mean():.3f}  highest {x.max():.3f}  margin to 0.5 before {0.5 - cl:.3f} after {0.5 - x[80:].mean():.3f}")
```

```python
n = np.arange(1, 121)
fig, ax = plt.subplots(2, 1, figsize=(9, 5), sharex=True, gridspec_kw={"height_ratios": [2, 1]})
ax[0].axvspan(1, 60, color="#cfe6e1", alpha=0.5)
ax[0].plot(n, x, "-o", ms=2.5, lw=0.8, color="#2f7f72")
for y, style in ((cl, "-"), (ucl, "--"), (lcl, "--")): ax[0].axhline(y, ls=style, color="#0b4f45", lw=1)
ax[0].axhline(0.5, color="#b23a2e", lw=1, label="specification 0.5")
ax[0].plot(beyond, x[np.array(beyond) - 1], "o", color="#b23a2e"); ax[0].axvline(eighth, color="#c27c0e", ls=":", label="eighth in a row")
ax[0].set_ylabel("haze 90 (EBC)"); ax[0].legend(fontsize=8)
ax[1].plot(n[1:], np.abs(np.diff(x)), lw=0.8, color="#2f7f72"); ax[1].axhline(mr_ucl, ls="--", color="#0b4f45")
ax[1].set_ylabel("moving range"); ax[1].set_xlabel("result (one per bright tank fill)"); plt.show()
```

**What to look at.** The chart flags the shift from result 88 while every result still passes the 0.5 specification. Change `base = x[:60]` to `x[:40]` as in the no code exercise and see how much the limits move.
