# Chapter 7: the water check beside the beer chart, then a gauge study
The density meter steps by 0.0002 g/cm3 at result 31 (seed 71, as in Figure 7.1). On beer that is 0.39 sigma and lost in the scatter; on the water check it is 20 sigma. Then a bitterness gauge study: 3 analysts x 5 retained samples x 2 readings (seed 72).

**No code tier.** `sheets/ch07-gauge-study.xlsx` splits the 30 readings into repeatability and the between analyst and day part with live formulas. For the charts: plot the first 30 water checks, draw 3 x sigma lines from the first 20, then add the rest.

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
DATA = "../data/"
```

```python
c = pd.read_csv(DATA + "ch07_checks.csv")
fig, ax = plt.subplots(2, 1, figsize=(9, 5), sharex=True)
for a, col, centre, sd, lab in ((ax[0], "beer_ae_plato", 2.20, 0.129, "beer AE (degrees Plato)"), (ax[1], "water_check_g_cm3", 0.99820, 0.00001, "water check (g/cm3)")):
    a.plot(c.result, c[col], "-o", ms=2.5, lw=0.8, color="#2f7f72")
    for y in (centre - 3 * sd, centre + 3 * sd): a.axhline(y, ls="--", color="#6b6b6b")
    a.axvline(30.5, color="#c27c0e", ls=":"); a.set_ylabel(lab)
ax[1].set_xlabel("result"); plt.show()
print("water step:", round(c.water_check_g_cm3[30:].mean() - c.water_check_g_cm3[:30].mean(), 5), "g/cm3")
```

```python
g = pd.read_csv(DATA + "ch07_gauge_study.csv")
gm = g.ibu.mean()
ss_a = 10 * ((g.groupby("analyst").ibu.mean() - gm) ** 2).sum()
ss_s = 6 * ((g.groupby("sample").ibu.mean() - gm) ** 2).sum()
ss_e = ((g.ibu - g.groupby(["analyst", "sample"]).ibu.transform("mean")) ** 2).sum()
ss_as = ((g.ibu - gm) ** 2).sum() - ss_a - ss_s - ss_e
ms_a, ms_as, ms_e = ss_a / 2, ss_as / 8, ss_e / 15
rep = math.sqrt(ms_e)
between = math.sqrt(max(0, (ms_a - ms_as) / 10) + max(0, (ms_as - ms_e) / 2))
within = math.hypot(rep, between)
print(f"repeatability sd {rep:.2f}  between analyst and day sd {between:.2f}  within lab sd {within:.2f}")
print(f"precision to tolerance {6 * within / 4:.2f} (repeatability alone {6 * rep / 4:.2f}); a common ceiling is 0.30")
```

**What to look at.** Read the two charts with the table in 7.1: the water check steps and the beer chart does not, so the instrument moved. The gauge study gives back the table in 7.2: bitterness uses 90 percent of its band.
